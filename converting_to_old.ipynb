```python
# Unnamed: 0	Owner Name	Property Location	Municipality	Search	Latitude	Longitude
#Owner Name	Property Location	Municipality	Search	County	PropertyLocationStrip	Rental	Latitude	Longitude
os.listdir('County_Data')
for file in os.listdir('County_Data'):
    df = pd.read_csv(f'County_Data/{file}')
    towns = df['Municipality'].unique()
    for town in towns:
        os.makedirs(f'Data_By_Towns_Index/{file[:-9]}', exist_ok=True)
        df_town = df[(df['Municipality'] == town) & (df['Search'].isna())]
        df_town.drop(columns=['County', 'PropertyLocationStrip'], inplace=True)
        df_town.reset_index(inplace=True)
        df_town.to_excel(f'Data_By_Towns_Index/{file[:-9]}/{town}.xlsx', index=False)

    
```
